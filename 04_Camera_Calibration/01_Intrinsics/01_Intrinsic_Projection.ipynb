# Camera Calibration - Intrinsic Parameters

## Objective
Understand and implement the camera intrinsic model.

##Topics
1. Camera intrinsic matrix (K)
2. perspective Projection
3. Back-Propagation
4. Projection verification

Libraries: Numpy, OpenCv, Matplotlib

```python
#Import libraries
import numpy as np
import cv2
import matplotlib.pyplot as plt

print("Numpy: ", np.__version__)
```

```python
#Define camera intrinsics


#Camera Parameters

fx = 600
fy = 600

cx = 320
cy = 240

# intrinsics matrix
K = np.array([
    [fx, 0, cx],
    [0, fy, cy],
    [0, 0, 1]
], dtype=float)

print("Intrinsic Matrix k: ")
print(K)
```

```python
#Impelement projection

def project_3d_to_2d(points_3d, K):

  points_h = np.array(points_3d, dtype=float)

  pixel_h = K @ points_h

  u = pixel_h[0] / pixel_h[2]
  v = pixel_h[1] / pixel_h[2]

  return np.array([u,v])

#Test Point
P = np.array([0.3, 0.2, 1.5])

pixel = project_3d_to_2d(P, K)

print("3D Point: ", P)
print("Image Pixel: ", pixel)
```

```python
#Implement back-projection

def back_project_2d_to_3d(pixel, depth, K):
  u, v = pixel
  cx = K[0,2]
  cy = K[1,2]

  X = (u - cx) * depth / fx
  Y = (v- cy) * depth / fy
  Z = depth

  return np.array([X, Y, Z])

recovered_point = back_project_2d_to_3d(
    pixel, P[2], K)

print("Orignal: ", P)
print("Recovered: ", recovered_point)

print("Verfication: ",
      np.allclose(P, recovered_point))
```
