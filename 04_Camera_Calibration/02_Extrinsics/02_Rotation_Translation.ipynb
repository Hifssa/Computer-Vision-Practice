## Part 1: Identity Rotation

The identity matrix presents zero rotation.
It preserve the orignal coordinates of a 3D point.represents.

```python
import numpy as np

#Define the i dentity rotation matrix

R = np.eye(3)

#Define a 3D point
P = np.array([2, 3, 4])

#Apply rotation
P_rotated = R @ P

print("Identity matrix: ", R)
print("orignal Point: ", P)
print("After Rotation: ", P_rotated)
```

```python
# Rotation around X_axis by 90 degrees

theta = np.radians(90)

Rx = np.array([
    [1, 0, 0],
    [0, np.cos(theta), -np.sin(theta)],
    [0, np.sin(theta), np.cos(theta)]
])

P = np.array([2, 3, 4])

P_rotated = Rx @ P

print("Rotation matrix:\n", Rx)
print("Orignal point", P)
print("Rotate point:", P_rotated)
```

```python
# Rotation arund y-axis by 90 degrees
theta = np.radians(90)

Ry = np.array([
    [np.cos(theta), 0, np.sin(theta)],
    [0,1,0],
    [-np.sin(theta), 0, np.cos(theta)]
])

P = np.array([2, 3, 4])

P_rotated = Ry @ P

print("Rotation matrix:/n", Ry)
print("orignal Point: ", P)
print("rotated point: ", P_rotated)
```

```python
#Rotation around Z_axis by 90 degrees

theta = np.radians(90)

Rz = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta), np.cos(theta), 0],
    [0, 0, 1]
])

P = np.array([2, 3, 4])

P_rotated = Rz @ P

print("Rotation Matrix for Z_axis:/n", Rz)
print("orignal Point: ", P)
print("rotated Point: ", P_rotated)
```

#Now we move to something important for camera calibration: combining multiple rotations into one rotation matrix.

Suppose we want to:
1. Rotate a point around X by 30°.
2. Then around Y by 45°.
3. Finally around Z by 90°.
Instead of applying three matrices separately, we can multiply them:
\[
R=R_z(90^\circ)R_y(45^\circ)R_x(30^\circ)
\]

Important: Matrix multiplication works right to left. Therefore, \(R_x\) acts first.

```python
import numpy as np
def rotation_x(angle):
  a = np.radians(angle)
  return np.array([
      [1, 0, 0],
      [0, np.cos(angle), -np.sin(angle)],
      [0, np.sin(angle), np.cos(angle)]
  ])

def rotation_y(angle):
  a = np.radians(angle)
  return np.array([
      [np.cos(angle), 0, np.sin(angle)],
      [0, 1, 0],
      [-np.sin(angle), 0, np.cos(angle)]
  ])

def rotation_z(angle):
  a = np.radians(angle)
  return np.array([
      [np.cos(angle), -np.sin(angle), 0],
      [np.cos(angle), np.sin(angle), 0],
      [0, 0, 1]
  ])

# Give angles
Rx = rotation_x(30)
Ry = rotation_y(45)
Rz = rotation_z(90)


#Combine rotations
R = Rz @ Ry @ Rx  #Matrix multiplication works right to left. Therefore, Rx acts first.
P = np.array([1, 2, 3])

P_rotated = R @ P

print("Combined rotation matrix:\n", R)
print("Original point:", P)
print("Rotated point:", P_rotated)
```

One important concept: rotation order
In general:
         RyRx :notequal RxRy

Rotating around X and then Y does not produce the same result as rotating around Y and then X.

This is called non-commutativity.

```python
R_reverse = Rx @ Ry @ Rz

P_reverse = R_reverse @ P

print("Original order:", P_rotated)
print("Reverse order:", P_reverse)
```

#Combining rotation and translation

This is the central equation for our camera-to-camera transformation:
$$\boxed{P_t=RP_d+t}
$$
Where:
- (P_d\): 3D point in the depth-camera coordinate system.
- \(R\): rotation matrix.
- \(t\): translation vector.
- \(P_t\): same point expressed in thermal-camera coordinates.
We rotate first, then translate.




```python
#Combine rigid transformation

P_depth = np.array([2, 3, 4])

R = rotation_z(90)

t = np.array([1, 2, 3])

P_thermal = R @ P_depth + t

print("Depth camera point:", P_depth)
print("Rotation matrix:\n", R)
print("Translation:", t)
print("Thermal camera point:", P_thermal)
```

#Transforming multiple 3D points
In research, we rarely transform just one point. We work with entire point clouds.

For example, a depth camera may produce thousands of 3D points representing a goat. We need to transform all of them into the thermal-camera coordinate system.

```python
# Three 3D points from the depth camera

P_depth = np.array([
    [2, 3, 4],
    [1, 2, 3],
    [4, 1, 2]
])

R = rotation_z(90)

t = np.array([1, 2, 3])

#Transform all points

P_thermal = (R @ P_depth.T).T + t

print("Depth points:\n", P_depth)
print("\nThermal points:\n", P_thermal)
```

#Understand this expression
$$(R @ P_depth.T).T + t$$
Our rotation matrix expects each point as a column vector. Therefore, we transpose the point array, rotate it, and transpose it back.
Original point array:
$$
P=
\begin{bmatrix}
x_1&y_1&z_1\\
x_2&y_2&z_2\\
x_3&y_3&z_3
\end{bmatrix}
$$
After transpose:
$$
P^T=
\begin{bmatrix}
x_1&x_2&x_3\\
y_1&y_2&y_3\\
z_1&z_2&z_3
\end{bmatrix}
$$
Now each point is represented as a column, so we can apply:
$$
RP^T
$$
Then transpose back to restore the original row-per-point format.

##One important distinction: the transpose is needed because of our chosen matrix convention. It is not compulsory in every implementation; row-vector formulations can avoid it.




```python
print("P_depth shape:", P_depth.shape)
print("P_depth.T shape:", P_depth.T.shape)
print("R shape:", R.shape)
print("Transformed shape:", P_thermal.shape)
```

#Does rotation and translation change distances?
##A rigid transformation changes a point cloud's position and orientation, but it must preserve its shape and distances.
In linear algebra, the length of a vector is called its norm, so subtracting two points gives a connecting vector and np.linalg.norm measures its length.

 In 3D space, the straight-line distance \(d\) between point \(P_1 = (x_1, y_1, z_1)\) and point \(P_2 = (x_2, y_2, z_2)\) uses the 3D Pythagorean theorem:
$$d=\sqrt{(x_{1}-x_{2})^{2}+(y_{1}-y_{2})^{2}+(z_{1}-z_{2})^{2}}$$


• np.linalg: This is NumPy's Linear Algebra sub-module. It contains standard matrix and vector math functions.
• .norm(...): Short for vector norm (magnitude or length). By default, it calculates the L2 norm, which is the standard straight-line Euclidean distance.
• P1 - P2: Subtracts the coordinates of P2 from P1 element-by-element to create a difference vector [(x1 - x2), (y1 - y2), (z1 - z2)].

```python
#Calculate the distantce between two #D points

def Calculate_distance(P1, P2):
  return np.linalg.norm(P1- P2)  # computes the Euclidean distance

#Orignal Points
P1 = P_depth[0]
P2 = P_depth[1]

#Transformed points
Q1 = P_thermal[0]
Q2 = P_thermal[1]

original_distance = Calculate_distance(P1, P2)
transformed_distance = Calculate_distance(Q1, Q2)
print("Original distance:", original_distance)
print("Transformed distance:", transformed_distance)
print("Distance preserved:",
      np.allclose(original_distance, transformed_distance))
```

```python
#Final Practical test and verify that implementation preserves the shape even when we use different rotation and translation

P = np.array([
    [1, 2, 3],
    [4, 2, 1],
    [2, 5, 3],
    [3, 1, 6]
])

R = rotation_y(45) @ rotation_x(30)
t = np.array([5, -2, 3])

Q = (R @ P.T).T + t

# Compare all the pairwise distances
for i in range(len(P)):
  for j in range(i +1, len(P)):
    d_before = np.linalg.norm(P[i]- P[j])
    d_after = np.linalg.norm(Q[i] - Q[j])

    print(f"Points {i+1}- {j+1}: ",
    np.allclose(d_before, d_after))
```
